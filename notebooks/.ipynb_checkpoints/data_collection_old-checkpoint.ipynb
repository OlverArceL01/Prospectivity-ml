## Extracción de Agua

```python
import pandas as pd
import numpy as np
```

```python
df2 = pd.read_excel("https://datos.gob.cl/dataset/715a929a-c7e6-4f25-ac27-bdcfd19a4890/resource/31fa612b-f3fa-496f-b4c1-c4d44453a719/download/extraccion-de-agua-segun-fuente-2019.xlsx")
df2 = df2.rename(columns={
            "Año": "year", 
            "Aguas Superficiales (1) / Surface water": "surface_water_ls",
            "Aguas Subterráneas (2) / Groundwater": "groundwater_ls",
            "Aguas adquiridas a terceros (3) / Purchased water": "purchased_water_ls",
            "Aguas de mar (4) / Seawater": "seawater_ls"})
df2.head(20)
```

```python
df3 = df2.copy()
columns = ["surface_water_ls", "groundwater_ls", "purchased_water_ls", "seawater_ls"]
for col in columns:
    df3[col[:-2] + "m3s"] = df3[col] * (1/1000) 
df3 = df3.drop(columns, axis=1)
df3.head(20)
```

```python
water2019 = df3[["surface_water_m3s", "groundwater_m3s", "purchased_water_m3s", "seawater_m3s"]].iloc[10, :]
percentages = (water2019 * 100 / water2019.sum()).round(2).astype(str) + "%"

print(percentages)
```

```python
url = "https://arcgisawa.sernageomin.cl/server/rest/services/VIEW_WGS84/FeatureServer/2"
layer = FeatureLayer(url)

df = layer.query(
    where="1=1",
    out_fields="*",
    return_geometry=True,
    as_df=True
)

print(df.shape)
print(df.head())
```

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    df,
    geometry="SHAPE",
    crs="EPSG:4326"
)

gdf.to_parquet(
    "instalaciones.parquet",
    index=False
)
```

# Geoquimica

```python
url = "https://services1.arcgis.com/OyjvVdFTl5hfSdX3/arcgis/rest/services/Compilaci%C3%B3n_de_Geoqu%C3%ADmica_de_sedimentos_de_Chile_WFL1/FeatureServer/0"

layer = FeatureLayer(url)

print(layer.properties.name)
print(layer.properties.maxRecordCount)
```

```python
df = layer.query(
    where="1=1",
    out_fields="*",
    return_geometry=True,
    as_df=True
)

print(df.shape)
print(df.info())
```

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    df,
    geometry="SHAPE",
    crs="EPSG:4326"
)

gdf.to_parquet(
    "geoquimica_muestra_en_analisis.parquet",
    index=False
)
```

```python
url = "https://services1.arcgis.com/OyjvVdFTl5hfSdX3/arcgis/rest/services/Compilaci%C3%B3n_de_Geoqu%C3%ADmica_de_sedimentos_de_Chile_WFL1/FeatureServer/1"

layer = FeatureLayer(url)

print(layer.properties.name)
print(layer.properties.maxRecordCount)

df = layer.query(
    where="1=1",
    out_fields="*",
    return_geometry=True,
    as_df=True
)

print(df.shape)
print(df.info())
gdf = gpd.GeoDataFrame(
    df,
    geometry="SHAPE",
    crs="EPSG:4326"
)

gdf.to_parquet(
    "geoquimica_muestra_publicada.parquet",
    index=False
)
```
